# ジャストインタイムコンパイル

jax.jit()はjaxコードを高速化するjitコンパイラ

```python
import os
os.environ["XLA_PYTHON_CLIENT_PREALLOCATE"] = "false"
os.environ["CUDA_VISIBLE_DEVICES"] = "0"   # 計算を画面表示に使っていない GPU 0 に寄せる場合
```

```python
import jax
import jax.numpy as jnp
jax.devices()
```

```python
def selu(x, alpha=1.67, lambla_=1.05):
    return lambla_ * jnp.where(x > 0, x, alpha * jnp.exp(x) - alpha)

x = jnp.arange(1000000)

%timeit selu(x).block_until_ready()
```

```python
def log2_if_rank_2(x):
    if x.ndim == 2:
        return jnp.log(x) / jnp.log(2.0)
    else:
        return x

jax.jit(log2_if_rank_2).trace(jnp.array([1, 2, 3])).jaxpr
```
